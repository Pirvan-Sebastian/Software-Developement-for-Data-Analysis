```python
# toti pinguinii sub 4kg sunt fete restul baieti - V1

# loc -- in liniile in care se respecta conditie, valoarea inlocuita
penguins.loc[penguins["body_mass_g"] <= 4000,"sex"]="Female"
penguins.loc[penguins["body_mass_g"] > 4000,"sex"]="Male"

penguins.head()
```

```python
# toti pinguinii sub 4kg sunt fete restul baieti - V2

penguins["sex"]=np.where(
    penguins["sex"].isna(),
    np.where(
        penguins["body_mass_g"]<4000,
        "Female",
        "Male"),
    penguins["sex"])

penguins.head()
```
